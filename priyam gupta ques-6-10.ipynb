```python
def is_palindrome(string:str)->bool:
    """This function checks a string is palindrome or not.
    Args:
        string: must be str value.
    Returns:
        This function returns a boolean value.
    Raises:
        TypeError: Input must a string only.
        ValueError: Empty string not allowed."""
    if not isinstance(string,str):
        raise TypeError ("input must be a string only.")
    if string.lower() == "":
        raise ValueError ("Empty string not allowed.")
    else:
        string =  string.lower().strip()
    rev_str = ""
    for char in string.lower():
        rev_str = char + rev_str
    return rev_str == string
if __name__ == "__main__":
    try:
        user_input = input("Enter your string here: ").lower().strip()
        if is_palindrome(user_input):
            print(user_input,"is a palindrome")
        else:
            print(user_input,"is not a palindrome")
    except (TypeError,ValueError) as e:
        print(e)
    except Ecxception as e:
        print(e)
        
    
```

```python
def count_vowels(text:str)->int:
    
    """This function helps to find the numbers of vowels in a string.
    Args:
        text: Must be a string.
    Returns:
        This function returns a number of vowels in a string.
    Raises:
        TypeError: Input must be a string only."""
    
    if not isinstance(text,str):
        raise TypeError ("Input must be a string only.")
    vowels = "aeiou"
    count = 0
    for num in text:
        if num in vowels:
            count = count + 1
    return count
if __name__ == "__main__":
    try:
        user_input = input("Enter your string here: ")
        result = count_vowels(user_input)
        print(user_input,"contains only",result,"vowels")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
def sum_finder(n:int)->int:
    
    """This function helps to find the total of any given number.
    Args:
        n: must be an int value.
    Returns:
        This function returns total of given number in int.
    Raises:
        TypeError: Input must be a integer only.
        ValueError: Negative numbers are not allowed."""
    
    if not isinstance(n,int):
        raise TypeError("Input must be an integer only.")
    if n < 0:
        raise ValueError("Negative numbers are not allowed.")
    total = 0
    while n > 0:
        digit = (n % 10)
        total = total + digit
        n = n // 10
    return total
if __name__ == "__main__":
    try:
        user_input = int(input("Enter your number here: "))
        result = sum_finder(user_input)
        print("Total of",user_input,"is:",result)
    except (TypeError,ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
    
```
