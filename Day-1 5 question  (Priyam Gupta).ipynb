```python
# Ques-1 This function used to reverse a string.

def reverse_string(name:str)->str:
    """This function used for reverse a string.
    Args: 
        name: should be a str value
    Returns:
        This function returns a reverse string.
    Raises:
        TypeError: Input should be a string.
        """
    if not isinstance (name,str):
        raise TypeError("Input should be in character form ")
    rev_str = ""
    for char in name:
        rev_str = char+rev_str
    return rev_str
if __name__ == "__main__":
    try:
        user_input = input("Enter your string here: ")
        result = reverse_string(user_input)
        print(result)
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
# Ques-2 Find the largest number from a list.

def largest_num_finder(list1:list(float))->float:
        
        """This Funtion is used to find the largest number from given list.
        Args:
            list1: (list1:list(float))
        Return:
            This Function return a largest float value.
        Raises:
            TypeError: Input must be a list (float or int)
            ValueError: Empty list not allowed."""
    
        if not isinstance (list1,list):
            raise TypeError ("Input must be a list (float or int)")
        for num in list1:
            if not isinstance(num,(float,int)):
                raise TypeError("Input must be int or float")
        if not list1:
            raise ValueError("Empty list not allowed")
            
        largest = list1[0]
        for num in list1:
            if num>largest:
                largest = num
        return largest
    
if __name__ == "__main__":
        
        try:
            user_input = input("Enter a list of numbers separated by space: ")
            list1 = [float(item) for item in user_input.split()]
            result = largest_num_finder(list1)
            print(f"Largest number is: {result}")
            
        except (TypeError,ValueError) as e:
            print(e)
        except Exception as e:
            print(e)
```

```python
# Q3- This function helps to find a prime number.

def is_prime_number(number:int)->bool:
    """This function helps to find a prime number.
    Args:
        number:int
    Returns: 
        This function return a boolean value.
    Riases:
        TypeError: Number should be an integer.
        ValueError: Number should be 2 or greater than 2 """
    
    if not isinstance(number,int):
        raise TypeError("Number should be an integer")
    if number < 2:
        raise ValueError("Number should be 2 or greater than 2")
    for i in range(2,int(number**.5)+1):
        if number % i == 0:
            return False
            break
    else:
        return True
if __name__ == "__main__":
    try:
        user_input = input("Enter a number which you want to check prime or not: ")
        if not user_input.strip().isdigit():
            raise TypeError ("Input must be a digit")
        number = int(user_input)
        if is_prime_number(number):
            print(f"{number} is not a Prime Number")
        else:
            print(f"{number} is a Prime Number")
    except (TypeError,ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
# Q-5
def factorial_finder(num:int)->int:
    
    """This func tion helps to find factorial of a number.
    Args:
        num: should be an int value.
    Return:
        It returns factorial of the given number and the number is an integer.
    Raises:
        TypeError: Input should be an Integer value."""
    
    if not isinstance (num,int):
        raise TypeError ("Input should be an int value")
    factorial = 1
    for item in range (1,num+1):
        factorial = factorial * item
    return factorial
if __name__ == "__main__":
       
    try:
        user_input = int(input("Enter a number which you want to find factorial: "))
        result = factorial_finder(user_input)
        print(result)
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```
