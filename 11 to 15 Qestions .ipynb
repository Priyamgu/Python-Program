```python
# Q-11
def recursive_facto(n:int)->int:
    """This function is used to find the factorial of given number.
    Args:
        n: should be an int value.
    Returns:
        This function returns factorial of given number.
    Raises:
        TypeError: Input must be an int value.
        ValueError: Negative values are not allowed."""
    if not isinstance (n,int):
        raise TypeError ("Input must be an int value.")
    if n < 0:
        raise ValueError ("Negative values are not allowed.")
    if n == 0 or n == 1:
        return 1
    return n * recursive_facto(n-1)
if __name__ == "__main__":
    try:
        user_input = int(input("Enter your number here: "))
        result = recursive_facto(user_input)
        print(f"The factorial of {user_input} is:----> {result}")
    except (TypeError,ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
# Q-12
def count_char(string:str)->dict:
    """This function is used to find the count of same characters in a string.
    Args:
        string: must be a str value.
    Returns:
        This function returns a dictionary in which all characters list with their count.
    Raises:
        TypeError: Input must be a str value."""
    if not isinstance (string,str):
        raise TypeError ("Input must be a str value.")
    char_count = {}
    for char in string.lower().replace(" ",""):
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    return char_count
if __name__ == "__main__":
    try:
        user_input = input("Enter your string here: ").lower().replace(" ","")
        result = count_char(user_input)
        print(f"The string's characters count in this dictionary is: {result}")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
# Q-13
def is_anagram(string1:str,string2:str)->bool:
    """This function used to find given two strings are anagram or not.
    Args:
        string1: must be a str value.
        string2: must be a str value.
    Returns: 
        This function returns a boolean value if both strings are same (True) if not (False).
    Raises:
        TypeError: Input must be a string.
    """
    if not isinstance (string1,str) or not isinstance (string2,str):
        raise TypeError ("Input must be string.")
    string1 = string1.lower().replace(" ","")
    string2 = string2.lower().replace(" ","")
    if len(string1) != len(string2):
        return False
    total_char = {}
    for char in string1:
        if char in total_char:
            total_char[char] += 1
        else:
            total_char[char] = 1
    
    for char in string2:
        if char not in total_char:
            return False
            break
        total_char[char] = 1
        if total_char[char] < 0:
            return False
    else:
        return True
if __name__ == "__main__":
    try:
        user1 = input("Enter Your First string here: ").lower().replace(" ","")
        user2 = input("Enter Your Second string here: ").lower().replace(" ","")
        result = is_anagram(user1,user2)
        print(f"{user1} and {user2} are anagram.")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
# Q-14
def fibonacci_seq(n:int)->None:
    """This function is used for fibonacci sequence.
    Args:
        n: must be an integer.
    Returns:
        This function returns fibonacci sequence and the type of data is: None.
    Raises:
        TypeError: Input must be an int value.
        ValueError: Input must be greater than zero."""
    if not isinstance(n,int):
        raise TypeError ("Input must be an int value.")
    if n < 0:
        raise ValueError ("Input must be greater than zero.")
    a = 0
    b = 1
    for i in range(n):
        print(a, end = " ") 
        a,b = b,a+b
if __name__ == "__main__":
    try:
        user_input = int(input("Enter a number here: "))
        result = fibonacci_seq(user_input)
        print(result)
    except (TypeError, ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
# Q-15
def check_leap_year(year:int)->bool:
    """This function is used to find a year is a leap year or not.
    Args: 
        year: should be an int value.
    Returns:
        This function returns boolean value. If the year is Leap year so it's True or not so it's False.
    Raises:
        TypeError: Input must be an integer.
        ValueError: Input must be greater than 1582."""
    if not isinstance (year,int):
        raise TypeError("Input must be an integer.")
    if year < 1582:
        raise ValueError ("Input must be greater than 1582")
    if year > 1582 and (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False
if __name__ == "__main__":
    try:
        user_input = int(input("Enter your year here: "))
        result = check_leap_year(user_input)
        print(f"{user_input} is a leap year: {result}")
    except (TypeError,ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
    
```
